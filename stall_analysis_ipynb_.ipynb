```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140}
}
```

## TASK 1: SALES PERFORMANCE ANALYSIS

In this section, I will calculate the total sales, average daily sales, highest sales and lowest sales over the 14-day period. I will also identify the day and week when the highest and lowest sales occurred.

```python
# TASK 1: SALES PERFORMANCE ANALYSIS

# Calculate total sales
total_sales = sum(daily_sales)

# Calculate average daily sales
average_sales = total_sales / len(daily_sales)

# Calculate highest and lowest sales
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the positions of highest and lowest sales
highest_index = daily_sales.index(highest_sales)
lowest_index = daily_sales.index(lowest_sales)

# Identify the best and worst days
best_day = days[highest_index]
worst_day = days[lowest_index]

# Identify the week numbers
best_week = (highest_index // 7) + 1
worst_week = (lowest_index // 7) + 1

# Display results
print("TOTAL SALES ANALYSIS")
print("--------------------")
print(f"Total Sales: KES {total_sales:,}")
print(f"Average Daily Sales: KES {average_sales:,.2f}")
print(f"Highest Sales: KES {highest_sales:,}")
print(f"Best Day: {best_day}, Week {best_week}")
print(f"Lowest Sales: KES {lowest_sales:,}")
print(f"Worst Day: {worst_day}, Week {worst_week}")
```

## TASK 2: WEEKLY SALES GROWTH

In this section, I will create a function called week_totals() to calculate the total sales for each week. I will then calculate the percentage growth or decline in sales from Week 1 to Week 2.

```python
# TASK 2: WEEKLY SALES GROWTH

def week_totals(sales):
    """
    Calculates the total sales for Week 1 and Week 2.
    Returns both totals as a tuple.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])

    return week1, week2


# Calculate weekly totals
week1_total, week2_total = week_totals(daily_sales)

# Calculate weekly growth percentage
growth = ((week2_total - week1_total) / week1_total) * 100

# Display results
print("WEEKLY SALES ANALYSIS")
print("---------------------")
print(f"Week 1 Sales: KES {week1_total:,}")
print(f"Week 2 Sales: KES {week2_total:,}")
print(f"Week-on-Week Growth: {growth:.2f}%")
```

## TASK 3: DAILY SALES CLASSIFICATION

In this section, I will create a function called classify_day() to classify each day's sales as Strong, Normal or Weak based on the average daily sales. I will also count the number of days in each category.

```python
# TASK 3: DAILY SALES CLASSIFICATION

def classify_day(amount, average):
    """
    Classifies daily sales as Strong, Normal or Weak
    based on the average daily sales.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


# Create a dictionary to count each category
sales_categories = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

# Classify each day's sales
for day, amount in zip(days, daily_sales):
    category = classify_day(amount, average_sales)
    sales_categories[category] += 1
    print(f"{day}: KES {amount:,} - {category}")

# Display category counts
print("\nSALES CATEGORY SUMMARY")
print("----------------------")

for category, count in sales_categories.items():
    print(f"{category}: {count} days")
```

## TASK 4: WEEKDAY AND WEEKEND SALES COMPARISON

In this section, I will calculate the average sales for weekdays (Monday to Friday) and weekends (Saturday and Sunday). I will compare their performance and calculate the percentage difference.

```python
# TASK 4: WEEKDAY AND WEEKEND SALES COMPARISON

# Separate weekday and weekend sales
weekday_sales = []
weekend_sales = []

for day, amount in zip(days, daily_sales):
    if day in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(amount)
    else:
        weekend_sales.append(amount)

# Calculate averages
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

# Display results
print("WEEKDAY VS WEEKEND ANALYSIS")
print("---------------------------")
print(f"Weekday Average: KES {weekday_average:,.2f}")
print(f"Weekend Average: KES {weekend_average:,.2f}")
print(f"Percentage Difference: {percentage_difference:.2f}%")

if weekend_average > weekday_average:
    print("Weekend sales perform better than weekday sales.")
else:
    print("Weekday sales perform better than weekend sales.")# TASK 4: WEEKDAY AND WEEKEND SALES COMPARISON

# Separate weekday and weekend sales
weekday_sales = []
weekend_sales = []

for day, amount in zip(days, daily_sales):
    if day in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(amount)
    else:
        weekend_sales.append(amount)

# Calculate averages
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
percentage_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

# Display results
print("WEEKDAY VS WEEKEND ANALYSIS")
print("---------------------------")
print(f"Weekday Average: KES {weekday_average:,.2f}")
print(f"Weekend Average: KES {weekend_average:,.2f}")
print(f"Percentage Difference: {percentage_difference:.2f}%")

if weekend_average > weekday_average:
    print("Weekend sales perform better than weekday sales.")
else:
    print("Weekday sales perform better than weekend sales.")
```

## TASK 5: STOCK REORDER AND GROSS MARGIN ANALYSIS

In this section, I will analyse the available stock for each product. I will identify products that need restocking based on their reorder levels and calculate the gross margin percentage for each product.

```python
# TASK 5: STOCK REORDER AND GROSS MARGIN ANALYSIS

def products_to_reorder(stock):
    """
    Returns a list of products whose stock quantity
    is at or below their reorder level.
    """
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


# Identify products that need restocking
reorder_products = products_to_reorder(stock)

print("STOCK REORDER ANALYSIS")
print("----------------------")

for product in reorder_products:
    print(f"Restock: {product}")

# Calculate gross margin for every product
print("\nGROSS MARGIN ANALYSIS")
print("---------------------")

for product, details in stock.items():
    cost = details["cost"]
    price = details["price"]

    gross_margin = ((price - cost) / price) * 100

    print(f"{product}: {gross_margin:.2f}%")# TASK 5: STOCK REORDER AND GROSS MARGIN ANALYSIS

def products_to_reorder(stock):
    """
    Returns a list of products whose stock quantity
    is at or below their reorder level.
    """
    reorder_list = []

    for product, details in stock.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


# Identify products that need restocking
reorder_products = products_to_reorder(stock)

print("STOCK REORDER ANALYSIS")
print("----------------------")

for product in reorder_products:
    print(f"Restock: {product}")

# Calculate gross margin for every product
print("\nGROSS MARGIN ANALYSIS")
print("---------------------")

for product, details in stock.items():
    cost = details["cost"]
    price = details["price"]

    gross_margin = ((price - cost) / price) * 100

    print(f"{product}: {gross_margin:.2f}%")
```
